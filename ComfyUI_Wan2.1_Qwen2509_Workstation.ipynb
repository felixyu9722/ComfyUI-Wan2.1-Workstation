```python
# @title 🛠️ 步骤 1：安装 ComfyUI 核心环境、插件 (Manager / VHS / RIFE 插帧) 与加速依赖
import os
import subprocess

%cd /content
if not os.path.exists('/content/ComfyUI'):
    !git clone --depth 1 https://github.com/comfyanonymous/ComfyUI.git

%cd /content/ComfyUI
!pip install -q -r requirements.txt
!pip install -q sageattention
!apt-get update -qq && apt-get install -y -qq aria2

# 安装核心插件：ComfyUI-Manager、VideoHelperSuite (MP4导出)、Frame-Interpolation (RIFE插帧)
custom_nodes = [
    ("ComfyUI-Manager", "https://github.com/ltdrdata/ComfyUI-Manager.git"),
    ("ComfyUI-VideoHelperSuite", "https://github.com/Kosinkadink/ComfyUI-VideoHelperSuite.git"),
    ("ComfyUI-Frame-Interpolation", "https://github.com/Fannovel16/ComfyUI-Frame-Interpolation.git"),
]
for name, repo in custom_nodes:
    target = f"/content/ComfyUI/custom_nodes/{name}"
    if not os.path.exists(target):
        subprocess.run(["git", "clone", "--depth", "1", repo, target], check=True)
        req_file = os.path.join(target, "requirements.txt")
        if os.path.exists(req_file):
            subprocess.run(["pip", "install", "-q", "-r", req_file], check=False)

# 安装 cloudflared 公网隧道客户端
if not os.path.exists('/usr/local/bin/cloudflared'):
    !wget -q https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64 -O /usr/local/bin/cloudflared
    !chmod +x /usr/local/bin/cloudflared

print('✅ 步骤 1 完成：ComfyUI 核心环境、插件与加速组件已就绪！')
```

```python
# @title 📦 步骤 2：下载公共基础组件 (VAE / UMT5 文本编码器 / CLIP Vision) 与 文生视频 (T2V) 模型
import os
import subprocess

BASE_URL = "https://huggingface.co/Comfy-Org/Wan_2.1_ComfyUI_repackaged/resolve/main/split_files"
COMFY_MODELS = "/content/ComfyUI/models"

downloads_step2 = [
    (f"{BASE_URL}/vae/wan_2.1_vae.safetensors", f"{COMFY_MODELS}/vae", "wan_2.1_vae.safetensors"),
    (f"{BASE_URL}/text_encoders/umt5_xxl_fp8_e4m3fn_scaled.safetensors", f"{COMFY_MODELS}/text_encoders", "umt5_xxl_fp8_e4m3fn_scaled.safetensors"),
    (f"{BASE_URL}/clip_vision/clip_vision_h.safetensors", f"{COMFY_MODELS}/clip_vision", "clip_vision_h.safetensors"),
    (f"{BASE_URL}/diffusion_models/wan2.1_t2v_1.3B_bf16.safetensors", f"{COMFY_MODELS}/diffusion_models", "wan2.1_t2v_1.3B_bf16.safetensors"),
    (f"{BASE_URL}/diffusion_models/wan2.1_t2v_14B_fp8_e4m3fn.safetensors", f"{COMFY_MODELS}/diffusion_models", "wan2.1_t2v_14B_fp8_e4m3fn.safetensors"),
]

for url, out_dir, filename in downloads_step2:
    os.makedirs(out_dir, exist_ok=True)
    target_path = os.path.join(out_dir, filename)
    if os.path.exists(target_path):
        print(f"✔️ 已就绪: {filename}")
        continue
    print(f"⬇️ 正在高速下载: {filename} ...")
    subprocess.run(["aria2c", "--console-log-level=error", "-c", "-x", "16", "-s", "16", "-k", "1M", url, "-d", out_dir, "-o", filename], check=True)

print("✅ 步骤 2 完成：公共组件与文生视频 (T2V 1.3B & 14B) 模型已就绪！")
```

```python
# @title ⚙️ 步骤 6：一键部署全部 14 套预置工作流 & 开启本地打包与 Google Drive 自动备份
import os
import shutil
import subprocess
from PIL import Image, ImageDraw

WF_DIR = "/content/ComfyUI/user/default/workflows"
INPUT_DIR = "/content/ComfyUI/input"
os.makedirs(WF_DIR, exist_ok=True)
os.makedirs(INPUT_DIR, exist_ok=True)

# 1. 从本项目 GitHub 仓库自动拉取最新全套 14 个预置工作流（确保新环境开箱即用）
if not os.path.exists("/tmp/wf_repo"):
    subprocess.run(["git", "clone", "--depth", "1", "https://github.com/felixyu9722/ComfyUI-Wan2.1-Workstation.git", "/tmp/wf_repo"], check=False)
if os.path.exists("/tmp/wf_repo/workflows"):
    for f in os.listdir("/tmp/wf_repo/workflows"):
        if f.endswith(".json"):
            shutil.copy2(os.path.join("/tmp/wf_repo/workflows", f), os.path.join(WF_DIR, f))

# 2. 生成默认示例占位图、示例音频与示例控制视频，防止未上传自定义素材前报错
for fname, color, label in [
    ("person_example.png", (60, 90, 140), "Image 1: Upload Person"),
    ("clothes_example.png", (140, 70, 90), "Image 2: Upload Clothes"),
    ("wan2.1_flf2v_720_f16_start_image.png", (50, 110, 90), "Start Frame Image"),
    ("wan2.1_flf2v_720_f16_end_image.png", (110, 80, 50), "End Frame Image"),
    ("video_wan_vace_14B_ref2v_reference_image.jpg", (80, 80, 120), "Reference Image"),
    ("flux_dev_example.png", (70, 100, 130), "Upload Start Image"),
    ("wan2.1_fun_control_start_frame.png", (80, 110, 90), "Upload Character Image"),
    ("two_character_talking.png", (90, 80, 120), "Upload Speaker Image"),
    ("video_wan2.1_fun_camera_v1.1_1.3B_start_image.jpg", (100, 90, 70), "Upload Camera Start Image"),
]:
    fpath = os.path.join(INPUT_DIR, fname)
    if not os.path.exists(fpath):
        img = Image.new("RGB", (832, 480), color=color)
        draw = ImageDraw.Draw(img)
        draw.text((260, 220), label, fill=(255, 255, 255))
        img.save(fpath)

for audio_name in ["audio_speaker1_woman.mp3", "audio_speaker2_man.mp3"]:
    apath = os.path.join(INPUT_DIR, audio_name)
    if not os.path.exists(apath):
        subprocess.run(["ffmpeg", "-y", "-f", "lavfi", "-i", "sine=frequency=440:duration=3", "-ac", "1", "-ar", "16000", apath], stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL, check=False)

vpath = os.path.join(INPUT_DIR, "wan2.1_fun_control_control_video.mp4")
if not os.path.exists(vpath):
    subprocess.run(["ffmpeg", "-y", "-f", "lavfi", "-i", "testsrc=duration=3:size=512x512:rate=16", "-pix_fmt", "yuv420p", vpath], stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL, check=False)

workflows = sorted([f for f in os.listdir(WF_DIR) if f.endswith(".json")], key=lambda x: int(x.split("_")[0]) if x.split("_")[0].isdigit() else 99)
print(f"📂 已在 {WF_DIR} 部署以下 {len(workflows)} 套工作流：")
for wf in workflows:
    print(f"   ✔️ {wf}")

# 3. 打包本地备份 ZIP 并确保 Google Drive 自动同步守护进程在后台运行
shutil.make_archive("/content/ComfyUI_Workflows_Backup", "zip", WF_DIR)
if os.path.exists("/content/ComfyUI/output"):
    shutil.make_archive("/content/ComfyUI_Outputs_Backup", "zip", "/content/ComfyUI/output")

sync_script = '''import os, time, shutil
while True:
    try:
        if os.path.exists("/content/drive/MyDrive"):
            backup_root = "/content/drive/MyDrive/ComfyUI_Backup"
            for sub, src in [("output", "/content/ComfyUI/output"), ("workflows", "/content/ComfyUI/user/default/workflows")]:
                if os.path.exists(src):
                    dst = os.path.join(backup_root, sub)
                    shutil.copytree(src, dst, dirs_exist_ok=True)
    except Exception:
        pass
    time.sleep(15)
'''
with open("/content/gdrive_sync_daemon.py", "w") as f:
    f.write(sync_script)
subprocess.run(["pkill", "-f", "gdrive_sync_daemon.py"], check=False)
subprocess.Popen(["python3", "/content/gdrive_sync_daemon.py"], stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)

# 如需挂载 Google Drive 云盘自动同步，可取消下方两行注释：
# from google.colab import drive
# drive.mount('/content/drive')

print("\n✅ 步骤 6 完成：14 套工作流与自动备份服务已就绪！")
```

```python
# @title 🚀 步骤 7：启动 ComfyUI 全能工作站并显示公网登录链接（最后一步）
import subprocess
import time
import re
import socket

# 清理已有进程，确保端口干净
subprocess.run(["pkill", "-f", "python3 main.py"], check=False)
subprocess.run(["pkill", "-f", "cloudflared"], check=False)
time.sleep(1)

# 启动 ComfyUI 后台服务（开启 CORS 跨站许可 + Gzip 响应压缩 + SageAttention 推理加速）
comfy_log = open("/content/comfyui.log", "w")
subprocess.Popen(
    [
        "python3", "main.py",
        "--listen", "127.0.0.1",
        "--port", "8188",
        "--enable-cors-header",
        "--enable-compress-response-body",
        "--use-sage-attention"
    ],
    cwd="/content/ComfyUI",
    stdout=comfy_log,
    stderr=subprocess.STDOUT
)

print("⏳ 正在启动 ComfyUI 服务端 (端口 8188)...")
for _ in range(60):
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as s:
        if s.connect_ex(("127.0.0.1", 8188)) == 0:
            print("✅ ComfyUI 本地引擎已就绪！")
            break
    time.sleep(1)

# 启动 Cloudflare HTTP/2 高速公网隧道
cf_log = open("/content/cloudflared.log", "w")
subprocess.Popen(
    ["/usr/local/bin/cloudflared", "tunnel", "--protocol", "http2", "--url", "http://127.0.0.1:8188"],
    stdout=cf_log,
    stderr=subprocess.STDOUT
)

public_url = None
for _ in range(30):
    time.sleep(1)
    with open("/content/cloudflared.log", "r") as f:
        content = f.read()
    match = re.search(r"https://[a-zA-Z0-9-]+\.trycloudflare\.com", content)
    if match:
        public_url = match.group(0)
        break

print("\n" + "=" * 72)
print("🎉 ComfyUI 全能视频与图像工作站已成功启动！")
print(f"👉 请点击下方登录链接进入 WebUI: {public_url}")
print("=" * 72)
print("💡 进入后操作提示：点击界面最左侧 📁 (Workflows) 图标即可一键切换 01~14 号预置工作流！")
```
